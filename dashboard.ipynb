```python
# Now start visualization.
```

```python
import matplotlib.pyplot as plt
```

```python
# City vs Average Price
```

```python
city_price = df.groupby("City")["Tour_Price_INR"].mean().sort_values()

plt.figure(figsize=(10,6))

plt.barh(city_price.index, city_price.values)

plt.xlabel("Average Tour Price (INR)")
plt.ylabel("City")
plt.title("Average Tour Price by City")

plt.show()
```

```python
import seaborn as sns
```

```python
# Tour Type vs Price

sns.boxplot(
    data=df,
    x="Tour_Type",
    y="Tour_Price_INR"
)

plt.title("Tour Price Distribution by Tour Type")
plt.show()
```

```python
#City bookings chart
```

```python
city_bookings = (
    df.groupby("City")["Bookings"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

sns.barplot(
    x=city_bookings.values,
    y=city_bookings.index
)

plt.title("Top 10 Cities by Bookings")
plt.xlabel("Total Bookings")
plt.ylabel("City")

plt.show()
```

```python
# Correlation analysis
```

```python
numeric_columns = [
    "Duration_Days",
    "Tour_Price_INR",
    "Travelers",
    "Rating",
    "Bookings",
    "Hotel_Cost",
    "Flight_Cost",
    "Food_Cost",
    "Activities_Cost",
    "Local_Transport_Cost"
]

corr = df[numeric_columns].corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```
